```python
import pandas as pd

!git clone https://github.com/ChiaviniK/fancygeneration
```

```python
clientes = '/content/fancygeneration/perfil_clientes.csv'
vendas = '/content/fancygeneration/vendas_pedidos.csv'
produtos = '/content/fancygeneration/catalogo_produtos.csv'

df_clientes = pd.read_csv(clientes)
df_vendas = pd.read_csv(vendas)
df_produtos = pd.read_csv(produtos)
```

```python
#Junção das Tabelas

#Vendas + Produtos
df_produtos_vendas = pd.merge(
    df_vendas,
    df_produtos,
    left_on='sku_produto',
    right_on='sku_produto',
    how='inner')

#Vendas_Produtos + clientes
df_geral = pd.merge(
    df_clientes,
    df_produtos_vendas,
    left_on='id_cliente',
    right_on='id_cliente',
    how='inner'
)
display(df_produtos_vendas)
```

```python
#Junção das Tabelas + Calculo de Lucros

#Vendas + Produtos
df_produtos_vendas = pd.merge(
    df_vendas,
    df_produtos,
    left_on='sku_produto',
    right_on='sku_produto',
    how='inner')

#Vendas_Produtos + clientes
df_geral = pd.merge(
    df_clientes,
    df_produtos_vendas,
    left_on='id_cliente',
    right_on='id_cliente',
    how='inner'
)

df_geral['Lucro'] = (
    df_geral['preco_venda'] - df_geral['custo_producao']

)


display (df_geral)
```

```python
# Calcular o total de itens que tem na coluna quantidade e quantos há na coluna preço de venda
# realiza a soma de cada coluna mencionada
quantidade_total = df_geral['quantidade'].sum()
total_faturamento = df_geral['preco_venda'].sum()

# Realiza o ticket médio de todas as vendas independente do produto

ticket_medio_vendas = total_faturamento / quantidade_total

print (f'O total de itens é de {quantidade_total}')
print(f' O faturamento total é de R$ {total_faturamento:.2f}') # o 2f formata a quantidade de números decimais - os números depois da vírgula,
print(f' O ticket médio é de R$ {ticket_medio_vendas:.2f}')
```

```python
# Calculo para saber quantos clientes únicos tivemos

total_clientes_unicos = df_geral['id_cliente'].nunique() # a função serve para contar o número de elementos únicos (distintos) em uma coluna ou dataframe.
print(f' o total de clientes unicos é de {total_clientes_unicos}')
```

```python
# Vamos saber o ticket médio por cliente.
ticket_medio_cliente = total_faturamento / total_clientes_unicos
print(f' O ticket médio por cliente é de R$ {ticket_medio_cliente:.2f}')
```

```python
# Para efetuar o download da planilh atualizada

df_geral.to_csv('Dados atualizados.csv',index = False) # index false f - você está dizendo ao Pandas para não salvar o índice das linhas no arquivo CSV final.
```
